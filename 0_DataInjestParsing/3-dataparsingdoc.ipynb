### Work document processing

```python
from docx import Document as doc
import os
```

```python
from langchain_community.document_loaders import Docx2txtLoader, UnstructuredWordDocumentLoader
```

```python
## Method 1: using DocxtxtLoader

print("Using Docx2txtLoader")
try:
    docx_loader = Docx2txtLoader("data/word_files/hello_stability_CTD_Report.docx")
    docs = docx_loader.load()
    print(docs)
    print(f"Loaded {len(docs)} documents")
    print(f"Content preview: {docs[0].page_content[:200]}....")
    print(f"Metadata: {docs[0].metadata}....")
except Exception as e:
    print(f"error: {e}")
```

```python
## Method 2 UnstructuredWordDocumentLoader
print(f"Using UnstructuredWordDocumentLoader")
try:
    unstructured_loader= UnstructuredWordDocumentLoader("data/word_files/hello_stability_CTD_Report.docx", mode="elements")
    unstructured_doc = unstructured_loader.load()
    print(f"Loaded {len(unstructured_doc)} elements")
    for i ,doc in enumerate(unstructured_doc[:3]):
        print(f"\n Element {i+1}")
        print(f"Type: {doc.metadata.get('category', 'unknown')}")
        print(f"Content: {doc.page_content[:100]}...")
    
except Exceptional as e:
    print(f"error: {e}")
```
